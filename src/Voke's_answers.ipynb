```python
import numpy as np

# voke's data: 7 days, 14 hours, 3 hubs
np.random.seed(42)

riders_needed = np.random.randint(5, 25, size=(7, 14, 3))

print("Riders needed shape:", riders_needed.shape)
print(riders_needed)
```

```python
hubs = np.array(['Ikeja', 'Yaba', 'Lekki'])

print(hubs)
```

```python
# Q1: Riders needed at the 95th percentile

all_hours = riders_needed.reshape(-1, 3)

p95 = np.percentile(all_hours, 95, axis=0)

# Round up to whole riders
p95_riders = np.ceil(p95).astype(int)

# Match rider requirements to hub names
p95_dict = dict(zip(hubs, p95_riders))

# Identify the hub needing the most riders
busiest_hub_idx = np.argmax(p95_riders)
busiest_hub = hubs[busiest_hub_idx]

print("\n=== BUSINESS Q1 ===")
print("Riders at 95th percentile:", p95_dict)
print(f"Most riders needed at: {busiest_hub}")
```

```python
# Voke's delivery orders
np.random.seed(42)

grid = np.random.randint(1, 15, size=(7, 14, 12))

print("Grid shape:", grid.shape)
```

```python
first_hour = 8

areas = np.array([
    'Ikeja', 'Yaba', 'Lekki', 'Victoria Island',
    'Surulere', 'Apapa', 'Ajah', 'Ikorodu',
    'Festac', 'Maryland', 'Ogba', 'Gbagada'
])
```

```python
import numpy as np

# Generate voke's one-way delivery times for 14 hours and 12 areas
np.random.seed(42)

one_way = np.random.uniform(15, 90, size=(14, 12))

print("Shape of one_way:", one_way.shape)
print(one_way)
```

```python
params = {"target_minutes": 60}
```

```python
# Q2: Busiest hour and longest trip

# Total orders across all days and areas for each hour
orders_per_hour = np.sum(grid, axis=(0, 2))

busiest_hour_idx = np.argmax(orders_per_hour)
busiest_clock_hour = first_hour + busiest_hour_idx

# Find the longest trip in each area
max_per_area = np.max(one_way, axis=0)

# Identify the area with the longest trip
longest_area_idx = np.argmax(max_per_area)
longest_area = areas[longest_area_idx]
longest_trip = max_per_area[longest_area_idx]

# Identify the hour of the longest trip
longest_hour = first_hour + np.argmax(
    one_way[:, longest_area_idx]
)

# Check whether all trips meet the target
within_target = np.all(
    one_way <= params['target_minutes']
)

print("\n=== BUSINESS Q2 ===")
print(
    f"Busiest hour: {busiest_clock_hour}:00 "
    f"with {orders_per_hour[busiest_hour_idx]:.0f} orders"
)

print(
    f"Longest one-way trip: {longest_area} "
    f"= {longest_trip:.1f} mins at {longest_hour}:00"
)

print(
    f"Within {params['target_minutes']} mins target? "
    f"{within_target}"
)
```

```python
import matplotlib.pyplot as plt
```

```python
# Chart 1: Riders needed per hub
plt.figure(figsize=(8, 5))
plt.bar(hubs, p95_riders)
plt.title("Riders Needed per Hub at 95th Percentile")
plt.xlabel("Delivery Hub")
plt.ylabel("Number of Riders")
plt.tight_layout()
plt.savefig("chart1_riders_p95.png")
plt.show()
plt.close()
```

```python
# Chart 2: Total orders by hour
plt.figure(figsize=(9, 5))

hours = np.arange(first_hour, first_hour + 14)

plt.plot(
    hours,
    orders_per_hour,
    marker='o'
)

plt.title("Total Orders by Hour of Day")
plt.xlabel("Hour of Day")
plt.ylabel("Total Orders")
plt.xticks(hours)
plt.grid(True)
plt.tight_layout()
plt.savefig("chart2_busiest_hour.png")
plt.show()
plt.close()
```

```python
# Chart 3: Longest one-way trip per area
plt.figure(figsize=(11, 5))

plt.bar(areas, max_per_area)

plt.title("Longest One-Way Trip per Area")
plt.xlabel("Delivery Area")
plt.ylabel("Travel Time (Minutes)")
plt.xticks(rotation=45)
plt.axhline(
    y=params['target_minutes'],
    linestyle='--',
    label='Target travel time'
)
plt.legend()
plt.tight_layout()
plt.savefig("chart3_longest_trip.png")
plt.show()
plt.close()
```
